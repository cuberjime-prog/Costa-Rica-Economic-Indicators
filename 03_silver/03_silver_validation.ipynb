```python
%sql
--- Validación Total de registros
SELECT COUNT(*) AS total_registros
FROM world_bank.costa_rica.silver_economic_indicators;
```

```python
%sql
--- Validación de Registros por indicador
SELECT
    indicator_name,
    COUNT(*) AS total_registros,
    MIN(year) AS primer_anio,
    MAX(year) AS ultimo_anio
FROM world_bank.costa_rica.silver_economic_indicators
GROUP BY indicator_name
ORDER BY indicator_name;
```

```python
%sql
--- Validación de Duplicados
SELECT
    country_code,
    year,
    indicator_code,
    COUNT(*) AS total
FROM world_bank.costa_rica.silver_economic_indicators
GROUP BY
    country_code,
    year,
    indicator_code
HAVING COUNT(*) > 1;
```

```python
%sql
--- Validación de Valores nulos
SELECT COUNT(*) AS valores_nulos
FROM world_bank.costa_rica.silver_economic_indicators
WHERE value IS NULL;
```
