```python
# Análisis de desempleo podremos analizar la evolución del desempleo e identificar los años con mayores y menores tasas.

from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Leer datos limpios
df = spark.table(
    "world_bank.costa_rica.silver_economic_indicators"
)

gold_unemployment = (
    df.filter(F.col("indicator_code") == "SL.UEM.TOTL.ZS")
    .select(
        "year",
        F.round("value", 2).alias("unemployment_pct")
    )
)

gold_unemployment.write.format("delta").mode("overwrite").saveAsTable(
    "world_bank.costa_rica.gold_unemployment_analysis"
)

display(gold_unemployment.orderBy("year"))
```
