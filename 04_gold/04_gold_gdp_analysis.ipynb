```python
# El indicador de World Bank ya representa el crecimiento porcentual anual del PIB real, por lo que no necesitamos calcularlo nuevamente.

from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Leer datos limpios
df = spark.table(
    "world_bank.costa_rica.silver_economic_indicators"
)

gold_gdp = (
    df.filter(F.col("indicator_code") == "NY.GDP.MKTP.KD.ZG")
    .select(
        "year",
        F.round("value", 2).alias("gdp_growth_pct")
    )
)

gold_gdp.write.format("delta").mode("overwrite").saveAsTable(
    "world_bank.costa_rica.gold_gdp_analysis"
)

display(gold_gdp.orderBy("year"))
```
