```python
%sql
--- El número real dependerá de la disponibilidad de cada indicador.
SELECT 'Economic Summary' AS tabla, COUNT(*) AS registros
FROM world_bank.costa_rica.gold_economic_summary

UNION ALL

SELECT 'GDP', COUNT(*)
FROM world_bank.costa_rica.gold_gdp_analysis

UNION ALL

SELECT 'Inflation', COUNT(*)
FROM world_bank.costa_rica.gold_inflation_analysis

UNION ALL

SELECT 'Unemployment', COUNT(*)
FROM world_bank.costa_rica.gold_unemployment_analysis

UNION ALL

SELECT 'Population', COUNT(*)
FROM world_bank.costa_rica.gold_population_analysis;
```

```python
%sql
--- Ahora podemos consultar la evolución económica reciente de Costa Rica:
SELECT
    year,
    ROUND(gdp_growth_pct, 2) AS gdp_growth_pct,
    ROUND(inflation_pct, 2) AS inflation_pct,
    ROUND(unemployment_pct, 2) AS unemployment_pct,
    CAST(population AS BIGINT) AS population
FROM world_bank.costa_rica.gold_economic_summary
WHERE year >= 2020
ORDER BY year DESC;
```
