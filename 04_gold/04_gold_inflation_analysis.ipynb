```python
# Análisis de inflación, Esta tabla nos permitirá identificar períodos de alta inflación, estabilidad y deflación.

from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Leer datos limpios
df = spark.table(
    "world_bank.costa_rica.silver_economic_indicators"
)

gold_inflation = (
    df.filter(F.col("indicator_code") == "FP.CPI.TOTL.ZG")
    .select(
        "year",
        F.round("value", 2).alias("inflation_pct")
    )
)

gold_inflation.write.format("delta").mode("overwrite").saveAsTable(
    "world_bank.costa_rica.gold_inflation_analysis"
)

display(gold_inflation.orderBy("year"))
```
