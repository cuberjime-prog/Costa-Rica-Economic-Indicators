```python
import requests
import json
from datetime import datetime, timezone


# Indicadores económicos de Costa Rica
indicadores = {
    "gdp_growth": "NY.GDP.MKTP.KD.ZG",
    "inflation": "FP.CPI.TOTL.ZG",
    "unemployment": "SL.UEM.TOTL.ZS",
    "population": "SP.POP.TOTL"
}

# Ubicación donde guardaremos los JSON
ruta = "/Volumes/world_bank/costa_rica/raw_data"

# Recorrer cada indicador
for nombre, codigo in indicadores.items():

    # Construir URL
    url = (
        f"https://api.worldbank.org/v2/country/CRI/"
        f"indicator/{codigo}"
    )

    # Consultar la API
    response = requests.get(
        url,
        params={
            "format": "json",
            "date": "2000:2025",
            "per_page": 100
        },
        timeout=30
    )

    # Validar la respuesta HTTP
    response.raise_for_status()

    # Convertir la respuesta a JSON
    data = response.json()

    # Validar que existan registros
    if not isinstance(data, list) or len(data) < 2:
        raise ValueError(f"Respuesta inesperada: {nombre}")

    if not isinstance(data[1], list):
        raise ValueError(f"Datos inválidos: {nombre}")

    # Crear archivo JSON
    archivo = f"{ruta}/{nombre}.json"

    # Guardar la respuesta original
    with open(archivo, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

    print(f"{nombre}: {len(data[1])} registros guardados")

print("Extracción finalizada correctamente")
```
